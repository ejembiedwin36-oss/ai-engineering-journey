```python
import torch

print("CUDA available:", torch.cuda.is_available())
```

```python
gpu = torch.cuda.get_device_properties(0)

print("GPU:", gpu.name)
print("GPU memory (GB):", round(gpu.total_memory / 1e9, 2))
```

```python
import time

size = 2000
a = torch.randn(size, size)
b = torch.randn(size, size)

start = time.perf_counter()
result = a @ b
cpu_time = time.perf_counter() - start

print("CPU time:", round(cpu_time, 4), "seconds")
```

```python
a_gpu = a.to("cuda")
b_gpu = b.to("cuda")

result_gpu = a_gpu @ b_gpu
torch.cuda.synchronize()  # Finish the warm-up calculation

start = time.perf_counter()
result_gpu = a_gpu @ b_gpu
torch.cuda.synchronize()  # Wait for the GPU to finish
gpu_time = time.perf_counter() - start

print("GPU time:", round(gpu_time, 4), "seconds")
print("Speedup:", round(cpu_time / gpu_time, 2), "times")
```
